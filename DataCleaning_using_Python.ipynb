```python
#Handling missing values
df=df.dropna(subset=['Customer_ID'])
```

```python
df.isnull().sum()
```

```python
df['Age'].unique()
```

```python
import re
# for i in df['Age'].apply(lambda x :re.findall('[0-9]',str(x))):
#     if len(i)<2:
#         print(i)
def extract_age(age):
    age_num=re.findall('[0-9]+',str(age))
    if len(age_num)>0:
        return age_num[0]
    else:
        return age

df['Age']=df['Age'].apply(lambda x: extract_age(x))
```

```python
df['Age'].unique()
```

```python
df_age=df[df['Age']!= 'nan years']['Age']
```

```python
df_age
```

```python
df.dropna()
```

```python
age_median=int(df_age.dropna().astype('int64').median())
```

```python
age_median
```

```python
df.replace('nan years',age_median,inplace=True)
```

```python
df['Age'].unique()
```

```python
df['Age']=df['Age'].fillna(age_median)
```

```python
df['Age'].unique()
```

```python
df['Purchase_Amount']=df['Purchase_Amount'].fillna(df['Purchase_Amount'].median())
```

```python
df.isnull().sum()
```

```python
df['Feedback_Score']=df['Feedback_Score'].fillna(df['Feedback_Score'].mode()[0])
```

```python
df.isnull().sum()
```

```python
for col in ['Gender','City','Country']:
    df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['Last_Purchase_Date']= df['Last_Purchase_Date'].ffill()
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

cols=['Age', 'Purchase_Amount', 'Feedback_Score']

plt.figure(figsize=(15,5))

for i,col in enumerate(cols,1):
    plt.subplot(1,3,i)
    sns.boxplot(y=df[col],color='skyblue',width=0.3)
    plt.title(f'Boxplot of {col}',fontsize=12)
    plt.ylabel('')
    plt.grid(axis='y', linestyle='--',alpha=0.6)

plt.tight_layout()
plt.show()
```
